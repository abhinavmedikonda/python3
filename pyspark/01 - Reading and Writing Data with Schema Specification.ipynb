```python
# Import necessary libraries
from pyspark.sql import SparkSession
from pyspark.sql.types import StructType, StructField, StringType, IntegerType

# Initialize Spark session
spark = SparkSession.builder \
    .appName("ReadCSVWriteJSON") \
    .getOrCreate()
```

```python
# Define schema for the CSV file
schema = StructType([
    StructField("id", IntegerType(), True),
    StructField("name", StringType(), True),
    StructField("age", IntegerType(), True),
    StructField("city", StringType(), True)
])
```

```python
# Read the CSV file
csv_df = spark.read.format("csv") \
    .option("header", "true") \
    .schema(schema) \
    .load("D:\\PySpark\\m1demo\\input_data.csv")

csv_df.show()
csv_df.printSchema()
```

```python
# Write the data to a JSON file
csv_df.write.format("json") \
    .mode("overwrite") \
    .save("D:\\PySpark\\m1demo\\output_data.json")
```

```python
# Verify the written JSON file
json_df = spark.read.format("json") \
    .load("D:\\PySpark\\m1demo\\output_data.json")

json_df.show()
json_df.printSchema()
```
