```python
# Import necessary libraries
from pyspark.sql import SparkSession
from pyspark.sql.functions import sum, avg, count

# Initialize Spark session
spark = SparkSession.builder \
    .appName("GroupAggregateDemo") \
    .getOrCreate()

# Create DataFrame directly
data = [
    (1, "John", 28, "New York", 3000),
    (2, "Emily", 22, "Los Angeles", 4000),
    (3, "Michael", 35, "Chicago", 5000),
    (4, "Sarah", 30, "Houston", 4000),
    (5, "David", 40, "Phoenix", 4500),
    (6, "Alice", 28, "New York", 3200),
    (7, "Bob", 25, "Los Angeles", 3500)
]

columns = ["id", "name", "age", "city", "salary"]

df = spark.createDataFrame(data, columns)

df.show()
df.printSchema()
```

```python
# Group by city and calculate aggregations
# Example 1: Calculate sum of salaries grouped by city
salary_sum_by_city = df.groupBy("city").agg(sum("salary").alias("total_salary"))
salary_sum_by_city.show()
```

```python
# Example 2: Calculate average age grouped by city
average_age_by_city = df.groupBy("city").agg(avg("age").alias("average_age"))
average_age_by_city.show()
```

```python
# Example 3: Count the number of people in each city
count_by_city = df.groupBy("city").agg(count("id").alias("people_count"))
count_by_city.show()
```

```python
# Example 4: Group by age and calculate average salary
average_salary_by_age = df.groupBy("age").agg(avg("salary").alias("average_salary"))
average_salary_by_age.show()
```
