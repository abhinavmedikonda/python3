```python
# Import necessary libraries
from pyspark.sql import SparkSession
from pyspark.sql.types import StructType, StructField, StringType, IntegerType

# Initialize Spark session
spark = SparkSession.builder \
    .appName("SelectFilterDemo") \
    .getOrCreate()
```

```python
# Create DataFrame directly
data = [
    (1, "John", 28, "New York"),
    (2, "Emily", 22, "Los Angeles"),
    (3, "Michael", 35, "Chicago"),
    (4, "Sarah", 30, "Houston"),
    (5, "David", 40, "Phoenix"),
    (6, "Alice", None, "New York"),
    (7, "Bob", 25, None)
]

schema = StructType([
    StructField("id", IntegerType(), True),
    StructField("name", StringType(), True),
    StructField("age", IntegerType(), True),
    StructField("city", StringType(), True)
])

df = spark.createDataFrame(data, schema)

df.show()
df.printSchema()
```

```python
# Select specific columns
selected_df = df.select("id", "name", "age")
selected_df.show()
```

```python
# Filter rows based on conditions
# Example 1: Filter rows where age > 30
filtered_df1 = df.filter(df["age"] > 30)
filtered_df1.show()
```

```python
# Example 2: Filter rows where city is 'New York'
filtered_df2 = df.filter(df["city"] == "New York")
filtered_df2.show()
```

```python
# Example 3: Filter rows where age is not null and age < 35
filtered_df3 = df.filter((df["age"].isNotNull()) & (df["age"] < 35))
filtered_df3.show()
```

```python
# Example 4: Apply multiple conditions (age > 25 and city is not null)
filtered_df4 = df.filter((df["age"] > 25) & (df["city"].isNotNull()))
filtered_df4.show()
```
