```python
# Import necessary libraries
from pyspark.sql import SparkSession

# Initialize Spark session
spark = SparkSession.builder \
    .appName("HandlingDuplicateDataDemo") \
    .getOrCreate()
```

```python
# Create a DataFrame with duplicate records
data = [
    (1, "John", 28, "New York"),
    (2, "Emily", 22, "Los Angeles"),
    (3, "Michael", 35, "Chicago"),
    (4, "Sarah", 30, "Houston"),
    (1, "John", 28, "New York"),  # Duplicate
    (2, "Emily", 22, "Los Angeles"),  # Duplicate
    (5, "David", 40, "Phoenix"),
    (6, "Alice", 29, "Chicago"),
    (3, "Michael", 35, "Chicago")  # Duplicate
]

columns = ["id", "name", "age", "city"]

df = spark.createDataFrame(data, columns)

print("Original DataFrame with Duplicates:")
df.show()
```

```python
# 1. Identify duplicate records based on all columns
print("Duplicate Records (Based on All Columns):")
duplicate_records = df.groupBy("id", "name", "age", "city").count().filter("count > 1")
duplicate_records.show()
```

```python
# 2. Remove duplicate rows (based on all columns)
print("DataFrame After Removing Duplicates (All Columns):")
deduplicated_df_all = df.dropDuplicates()
deduplicated_df_all.show()
```

```python
# 3. Remove duplicate rows based on specific columns (e.g., 'id' and 'name')
print("DataFrame After Removing Duplicates (Based on 'id' and 'name'):")
deduplicated_df_specific = df.dropDuplicates(["id", "name"])
deduplicated_df_specific.show()
```
