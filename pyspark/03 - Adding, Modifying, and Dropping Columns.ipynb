```python
# Import necessary libraries
from pyspark.sql import SparkSession
from pyspark.sql.functions import col, when

# Initialize Spark session
spark = SparkSession.builder \
    .appName("AddModifyDropColumnsDemo") \
    .getOrCreate()

# Create DataFrame directly
data = [
    (1, "John", 28, "New York"),
    (2, "Emily", 22, "Los Angeles"),
    (3, "Michael", 35, "Chicago"),
    (4, "Sarah", 30, "Houston"),
    (5, "David", 40, "Phoenix"),
    (6, "Alice", None, "New York"),
    (7, "Bob", 25, None)
]

columns = ["id", "name", "age", "city"]

df = spark.createDataFrame(data, columns)

df.show()
df.printSchema()
```

```python
# Add a derived column
# Add a new column `age_category` based on age
df_with_added_column = df.withColumn(
    "age_category",
    when(col("age") < 30, "Under 30")
    .when((col("age") >= 30) & (col("age") < 40), "30s")
    .otherwise("40+")
)
df_with_added_column.show()
```

```python
# Modify a column
# Rename the column `city` to `location`
df_with_renamed_column = df_with_added_column.withColumnRenamed("city", "location")
df_with_renamed_column.show()
```

```python
# Drop a column
# Drop the `age_category` column
df_with_dropped_column = df_with_renamed_column.drop("age_category")
df_with_dropped_column.show()
```
