```python
# Import necessary libraries
from pyspark.sql import SparkSession

# Initialize Spark session
spark = SparkSession.builder \
    .appName("SetOperationsDemo") \
    .getOrCreate()
```

```python
# Create first DataFrame (Dataset A)
data_a = [
    (1, "John", 28),
    (2, "Emily", 22),
    (3, "Michael", 35),
    (4, "Sarah", 30)
]

columns_a = ["id", "name", "age"]

df_a = spark.createDataFrame(data_a, columns_a)
print("Dataset A:")
df_a.show()
```

```python
# Create second DataFrame (Dataset B)
data_b = [
    (3, "Michael", 35),
    (4, "Sarah", 30),
    (5, "David", 40),
    (6, "Alice", 29)
]

columns_b = ["id", "name", "age"]

df_b = spark.createDataFrame(data_b, columns_b)
print("Dataset B:")
df_b.show()
```

```python
# Perform Set Operations
# 1. Union (Combine two datasets)
union_df = df_a.union(df_b)
print("Union:")
union_df.show()
```

```python
# 2. Intersection (Common rows between two datasets)
intersection_df = df_a.intersect(df_b)
print("Intersection:")
intersection_df.show()
```

```python
# 3. Difference (Rows in Dataset A but not in Dataset B)
difference_df = df_a.subtract(df_b)
print("Difference (A - B):")
difference_df.show()
```

```python
# Remove duplicates from Union result (optional if required for distinct dataset)
union_distinct_df = union_df.distinct()
print("Union with Distinct Rows:")
union_distinct_df.show()
```
