```python
# Import necessary libraries
from pyspark.sql import SparkSession

# Initialize Spark session
spark = SparkSession.builder \
    .appName("JoinOperationsDemo") \
    .getOrCreate()
```

```python
# Create first DataFrame (Employee details)
employee_data = [
    (1, "John", 28, "New York"),
    (2, "Emily", 22, "Los Angeles"),
    (3, "Michael", 35, "Chicago"),
    (4, "Sarah", 30, "Houston")
]

employee_columns = ["emp_id", "name", "age", "city"]

employee_df = spark.createDataFrame(employee_data, employee_columns)

# Display Employee DataFrames
print("Employee DataFrame:")
employee_df.show()
```

```python
# Create second DataFrame (Department details)
department_data = [
    (1, "HR"),
    (2, "Finance"),
    (3, "Engineering"),
    (5, "Marketing")
]

department_columns = ["emp_id", "department"]

department_df = spark.createDataFrame(department_data, department_columns)

# Display Department DataFrames
print("Department DataFrame:")
department_df.show()
```

```python
# Perform Join Operations
# 1. Inner Join
print("Inner Join:")
inner_join_df = employee_df.join(department_df, on="emp_id", how="inner")
inner_join_df.show()
```

```python
# 2. Left Join
print("Left Join:")
left_join_df = employee_df.join(department_df, on="emp_id", how="left")
left_join_df.show()
```

```python
# 3. Right Join
print("Right Join:")
right_join_df = employee_df.join(department_df, on="emp_id", how="right")
right_join_df.show()
```

```python
# 4. Full Outer Join
print("Full Outer Join:")
outer_join_df = employee_df.join(department_df, on="emp_id", how="outer")
outer_join_df.show()
```
